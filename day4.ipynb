```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 100)   # 100 evenly spaced x-values from -5 to 5

y1 = x**2                     # base parabola
y2 = x**2 + 3                 # shifted up 
y3 = -x**2                    # reflected (flipped)

plt.plot(x, y1, label="y = x^2")
plt.plot(x, y2, label="y = x^2 + 3")
plt.plot(x, y3, label="y = -x^2")

plt.legend()
plt.title("y = x^2 and transformations")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
```
