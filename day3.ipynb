```python
def f(x):
    return x**2
def g(x):
    return x+1
def fg(x):
    return f(g(x))
def gf(x):
    return g(f(x))

# test fg against hand calculation: f(g(x)) = (x+1)^2
print(fg(2))  # g(2)=3, f(3)=9
print(fg(0))  # g(0)=1, f(1)=1
print(fg(-3)) # g(-3)=-2, f(-2)=4

# compare with gf to see they differ
print(gf(2)) # f(2)=4, g(4)=5
```

```python
def share_price(EPS,PE):            # f(x): EPS, PE -> Price
    return EPS*PE
def implied_PE (PRICE, EPS):        # f^-1(x): Price, EPS -> PE

    if EPS == 0:
        return "undefined"          # domain restriction
    return PRICE/EPS

print(share_price(6.25,24))        # -> 150.0

print(implied_PE(150,6.25))        # -> 24.0
```
