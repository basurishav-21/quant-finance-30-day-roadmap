```python
def f(x):
    return 2*x + 3 
print(f(5))
print(f(0))
print(f(-2))
```

```python
#1. A simple linear function
def g(x):
    return 3*x-4
print(g(1), g(0), g(-1))
```

```python
#2. A function with two inputs
def add_and_double(a,b):
    return (a+b)*2
print(add_and_double(2,3), add_and_double(0,0), add_and_double(-1,5))
```

```python
#3. A quadratic
def h(x):
    return x**2 + 6*x + 9
print(h(-3), h(0), h(3))
```

```python
#4. A function with a domain restriction
def reciprocal(x):
    if x == 0:
        return "undefined"
    return 1/x
print(reciprocal(2), reciprocal(0), reciprocal(-4))
```

```python
#5. Finance X Math scope limitations 
def share_price(eps, PE):
    return eps*PE
print(share_price(25,18))  #normal_case
print(share_price(-5,18))  #negative_EPS: Mathematically_fine_financially_meaningless
print(share_price(25,-18)) #negative PE

#PE becomes meaningless if Earnings is negative because this makes the denominator of the ratio negative which should turn the ratio negative as well, in such situations ratios like PS and PCF (if CF +ve must be used)
#PE negative without earnings negative cannot occur because Price can never go below 0
```
